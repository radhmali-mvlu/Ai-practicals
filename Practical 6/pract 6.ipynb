```python
#  practical 6 A CODE:
# A Bayesian Network using pgmpy
# Example: Burglary-Alarm Network
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm"),
    ("Alarm", "JohnCalls"),
    ("Alarm", "MaryCalls")
])
# Step 2: Define Conditional Probability Tables (CPDs)
# P(Burglary)
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.99], [0.01]],
    state_names={"Burglary": ["False", "True"]}
)
# P(Earthquake)
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)
# P(Alarm | Burglary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],
        [0.001, 0.29, 0.94, 0.95]
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)
# P(JohnCalls | Alarm)
cpd_john = TabularCPD(
    variable="JohnCalls",
    variable_card=2,
    values=[
        [0.95, 0.10],
        [0.05, 0.90]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "JohnCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)
# P(MaryCalls | Alarm)
cpd_mary = TabularCPD(
    variable="MaryCalls",
    variable_card=2,
    values=[
        [0.99, 0.30],
        [0.01, 0.70]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "MaryCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)
# Step 3: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm,
    cpd_john,
    cpd_mary
)
# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())
# Step 5: Perform inference
inference = VariableElimination(model)
# Query: Probability of Burglary when John and Mary both call
result = inference.query(
    variables=["Burglary"],
    evidence={
        "JohnCalls": "True",
        "MaryCalls": "True"
    }
)
print("\nProbability of Burglary given JohnCalls=True and MaryCalls=True:")
print(result)
print("radhika 17")
```

```python
# Bayesian Network Example: Medical Diagnosis
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Disease", "Fever"),
    ("Disease", "Cough"),
    ("Disease", "Test")
])
# Step 2: Define Conditional Probability Tables (CPDs)
# P(Disease)
cpd_disease = TabularCPD(
    variable="Disease",
    variable_card=2,
    values=[[0.90], [0.10]],
    state_names={"Disease": ["False", "True"]}
)
# P(Fever | Disease)
cpd_fever = TabularCPD(
    variable="Fever",
    variable_card=2,
    values=[
        [0.80, 0.20],
        [0.20, 0.80]
    ],
    evidence=["Disease"],
    evidence_card=[2],
    state_names={
        "Fever": ["False", "True"],
        "Disease": ["False", "True"]
    }
)
# P(Cough | Disease)
cpd_cough = TabularCPD(
    variable="Cough",
    variable_card=2,
    values=[
        [0.70, 0.30],
        [0.30, 0.70]
    ],
    evidence=["Disease"],
    evidence_card=[2],
    state_names={
        "Cough": ["False", "True"],
        "Disease": ["False", "True"]
    }
)
# P(Test | Disease)
cpd_test = TabularCPD(
    variable="Test",
    variable_card=2,
    values=[
        [0.95, 0.10],
        [0.05, 0.90]
    ],
    evidence=["Disease"],
    evidence_card=[2],
    state_names={
        "Test": ["Negative", "Positive"],
        "Disease": ["False", "True"]
    }
)
# Step 3: Add CPDs to model
model.add_cpds(
    cpd_disease,
    cpd_fever,
    cpd_cough,
    cpd_test
)
# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())
# Step 5: Perform inference
inference = VariableElimination(model)
# Query: Probability of Disease when Fever and Cough are True
result = inference.query(
    variables=["Disease"],
    evidence={
        "Fever": "True",
        "Cough": "True"
    }
)
print("\nProbability of Disease given Fever=True and Cough=True:")
print(result)
print("radhika 17")
```

```python
# Posterior Probability using Bayesian Network
# Query: P(Burgary | Alarm = True)
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burgary", "Alarm"),
    ("Earthquake", "Alarm")
])
# Step 2: Define probability of Burgary
cpd_burgary = TabularCPD(
    variable="Burgary",
    variable_card=2,
    values=[[0.999], [0.001]],
    state_names={"Burgary": ["False", "True"]}
)
# Step 3: Define probability of Earthquake
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)
# Step 4: Define probability of Alarm
# P(Alarm | Burgary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],  # Alarm = False
        [0.001, 0.29, 0.94, 0.95]   # Alarm = True
    ],
    evidence=["Burgary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burgary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)
# Step 5: Add CPDs to model
model.add_cpds(
    cpd_burgary,
    cpd_earthquake,
    cpd_alarm
)
# Step 6: Check whether model is correct
print("Is Bayesian Network valid?", model.check_model())
# Step 7: Perform inference
inference = VariableElimination(model)
result = inference.query(
    variables=["Burgary"],
    evidence={"Alarm": "True"}
)
print("Posterior Probability:")
print("P(Burgary | Alarm = True)")
print(result)
print("radhika 17")
```

```python
# Bayesian Network Example: Medical Diagnosis
# Query: P(Flu | Fever = True, Cough = True)
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Flu", "Fever"),
    ("Flu", "Cough")
])
# Step 2: Define probability of Flu
cpd_flu = TabularCPD(
    variable="Flu",
    variable_card=2,
    values=[[0.9], [0.1]],
    state_names={"Flu": ["False", "True"]}
)
# Step 3: Define probability of Fever
# P(Fever | Flu)
cpd_fever = TabularCPD(
    variable="Fever",
    variable_card=2,
    values=[
        [0.95, 0.20],   # Fever = False
        [0.05, 0.80]    # Fever = True
    ],
    evidence=["Flu"],
    evidence_card=[2],
    state_names={
        "Fever": ["False", "True"],
        "Flu": ["False", "True"]
    }
)
# Step 4: Define probability of Cough
# P(Cough | Flu)
cpd_cough = TabularCPD(
    variable="Cough",
    variable_card=2,
    values=[
        [0.85, 0.30],   # Cough = False
        [0.15, 0.70]    # Cough = True
    ],
    evidence=["Flu"],
    evidence_card=[2],
    state_names={
        "Cough": ["False", "True"],
        "Flu": ["False", "True"]
    }
)
# Step 5: Add CPDs to model
model.add_cpds(
    cpd_flu,
    cpd_fever,
    cpd_cough
)
# Step 6: Check whether model is correct
print("Is Bayesian Network valid?", model.check_model())

# Step 7: Perform inference
inference = VariableElimination(model)

result = inference.query(
    variables=["Flu"],
    evidence={
        "Fever": "True",
        "Cough": "True"
    }
)
print("Posterior Probability:")
print("P(Flu | Fever = True, Cough = True)")
print(result)
print("radhika 17")
```
